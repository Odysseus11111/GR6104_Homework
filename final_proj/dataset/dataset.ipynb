```python
import yfinance as yf

tickers = ["AAPL", "TSLA", "KO", "WMT"]

df = yf.download(
    tickers,
    start="2025-01-01",
    end="2026-01-01",
    auto_adjust=True
)

prices = df["Close"]
prices = prices.dropna()

prices.to_csv("prices_4stocks.csv")

print(prices.head())
print(prices.shape)
```
